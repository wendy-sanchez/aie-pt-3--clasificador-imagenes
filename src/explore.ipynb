```python
import os
import matplotlib.pyplot as plt
from tensorflow.keras.preprocessing.image import load_img
```

```python
dog_folder = "../data/raw/dogs-vs-cats/train"

dog_images = [
    image for image in os.listdir(dog_folder)
    if image.startswith("dog.")
][:9]

plt.figure(figsize=(12, 8))

for i, image_name in enumerate(dog_images):
    img = load_img(os.path.join(dog_folder, image_name))

    plt.subplot(3, 3, i + 1)
    plt.imshow(img)
    plt.title("Dog")
    plt.axis("off")

plt.tight_layout()
plt.show()
```
